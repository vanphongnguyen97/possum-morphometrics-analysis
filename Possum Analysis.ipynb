# Step 2: Data preprocessing and cleaning

```python
df_replace=df.replace(['?','n/a','na','--',',','N/A','NaN',' '],np.nan)

# Replace "," to "."
cols=['age','hdlngth','skullw','totlngth','taill','footlgth','earconch','eye','chest','belly']
df_replace[cols]=df_replace[cols].replace(",",".",regex=True)
df_replace[cols]=df_replace[cols].apply(pd.to_numeric,errors='coerce')

# Replace negative values with NaN
df_replace[cols] = df_replace[cols].mask(df_replace[cols] < 0, np.nan)
```

# Step 3: Handling missing values

```python
df_replace=df_replace.fillna({
    'case': 95,
    'site': 7,
    'Pop': "other",
})
```

```python
print(df_replace['case'].unique())
print(df_replace['site'].unique())
print(df_replace['Pop'].unique())
print(df_replace['gender'].unique())
```

```python
# Calculate median based on group to evaluate.
numeric_cols = ['age', 'hdlngth', 'skullw', 'totlngth', 'taill',
                'footlgth', 'earconch', 'eye', 'chest', 'belly']
group_cols = ['Pop', 'gender']

# Compute median of each numeric column based on group
group_medians = df_replace.groupby(group_cols)[numeric_cols].median()

print(group_medians)
```

# interpretation: 
The medians of the numeric columns for each (Pop, gender) group are quite similar. For example, the median age ranges only from 3.6 to 4.2, hdlngth from 91.95 to 93.3, and belly from 32.0 to 33.50. The differences across groups are relatively small, indicating that **there is no substantial variation between subgroups**. Therefore, it is reasonable **not to use group-based median to fill missing value** and instead **use the overall median of each column**.

```python
# List of columns to fill missing values
cols_to_fill = ['age','hdlngth','skullw','totlngth','footlgth','eye','belly','taill','earconch','chest']

# Fill missing values with overall median
for col in cols_to_fill:
    median_value = df_replace[col].median()
    df_replace[col].fillna(median_value, inplace=True)

print(df_replace.isna().sum())
```

# Step 4: Outlier Detection and Treatment

```python
# Visualize boxplots before outlier treatment
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = ['age', 'hdlngth', 'skullw', 'totlngth', 'taill', 
                'footlgth', 'earconch', 'eye', 'chest', 'belly']

plt.figure(figsize=(20, 12))
for i, column in enumerate(numeric_cols, 1):
    plt.subplot(3, 4, i)
    sns.boxplot(y=df_replace[column])
    plt.title(f'Boxplot of {column} before Outlier Treatment')
plt.tight_layout()
plt.show()
```

```python
# List of numeric columns
numeric_cols = ['age', 'hdlngth', 'skullw', 'totlngth', 'taill', 
                'footlgth', 'earconch', 'eye', 'chest', 'belly']

df_fixed = df_replace.copy()


for col in numeric_cols:
    # Step 1: Calculate Q1, Q3, and IQR
    Q1 = df_fixed[col].quantile(0.25)
    Q3 = df_fixed[col].quantile(0.75)
    IQR = Q3 - Q1
    
    # Step 2: Calculate lower and upper bounds
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    # Step 3: Find overall median
    median_value = df_fixed[col].median()
    
    # Step 4: Create a new column to store cleaned values
    new_col = f"{col}_no_outlier"
    df_fixed[new_col] = df_fixed[col]
    
    # Step 5: Create mask for outliers
    mask = (df_fixed[col] < lower) | (df_fixed[col] > upper)
    
    # Step 6: Replace outliers with overall median
    df_fixed.loc[mask, new_col] = median_value

```

```python
# Check remaining outliers in the new columns
remaining_outliers = {}
for col in numeric_cols:
    new_col = f"{col}_no_outlier"
    Q1 = df_fixed[new_col].quantile(0.25)
    Q3 = df_fixed[new_col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outlier_count = ((df_fixed[new_col] < lower) | (df_fixed[new_col] > upper)).sum()
    remaining_outliers[new_col] = outlier_count

remaining_outliers
```

```python
print("==Remaining outliers checking and fixing==")
columns_to_fix = ['hdlngth_no_outlier', 'skullw_no_outlier']

for col in columns_to_fix:
    while True:
        # Compute global IQR
        Q1 = df_fixed[col].quantile(0.25)
        Q3 = df_fixed[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        
        # Detect outliers
        mask = (df_fixed[col] < lower) | (df_fixed[col] > upper)
        num_outliers = mask.sum()
        
        # If no outliers left, stop
        if num_outliers == 0:
            break
        
        # Replace outliers with the overall median of the column
        median_val = df_fixed[col].median()
        df_fixed.loc[mask, col] = median_val

# Check remaining outliers
remaining_outliers = {}
for col in columns_to_fix:
    Q1 = df_fixed[col].quantile(0.25)
    Q3 = df_fixed[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    remaining_outliers[col] = ((df_fixed[col] < lower) | (df_fixed[col] > upper)).sum()

print("Remaining outliers after cleaning:", remaining_outliers)
```

## Hypothesis 1  

only title of H0 & H1: 


H₀ (null): There is no significant difference in head length between males and females. H0​:μm​=μf

H₁ (alternative): There is a significant difference in head length between males and females.
H1​:μm​≠μf​


```python
from scipy.stats import ttest_ind
print("Hypothesis 1: Gender difference in head length")
print(f"\n Step 1: Formulate Hypotheses")
print("H₀ (null): H0​:μ_m ​= μ_f (There is no significant difference in head length between males and females)")
print("H₁ (alternative): H1​:μ_m ​≠ μ_f (There is a significant difference in head length between males and females)")


print(f"\n Step 2: Choose Significance Level")
alpha = 0.05 
print(f"Significance Level: α = {alpha}")

print(f"\n Step 3: Perform the T-Test")
male = df_final[df_final['gender'] == 'm']['hdlngth_no_outlier']
female = df_final[df_final['gender'] == 'f']['hdlngth_no_outlier']
t_stat, p_val = ttest_ind(male, female, equal_var=False)
print(f"T-statistic: {t_stat}, P-value: {p_val}")


print(f"\n Step 4: Make a Decision")
if p_val < alpha:
    print(f"✅ Reject the null hypothesis (p = {p_val:.4f}).")
    print("There is a statistically significant difference in head length between males and females.")
else:
    print(f"❌ Fail to reject the null hypothesis (p = {p_val:.4f}).")
    print("There is no statistically significant difference in head length between males and females.")
```

## Hypothesis 2

Only title of H0 & H1:

H0​: μ_Vic ​= μ_Other​ (There is no significant difference in skull width between populations.)

H1​: μ_Vic ​​≠ μ_Other (There is a significant difference in skull width between populations.)

```python
print("Hypothesis 2: Population difference in Skull width")
print(f"\n Step 1: Formulate Hypotheses")
print("H₀ (null): H0​:μ_Vic ​= μ_other (There is no significant difference in skull width between populations.)")
print("H₁ (alternative): H1​:μ_Vic ​≠ μ_other (There is a significant difference in skull width between populations.)")

print(f"\n Step 2: Choose Significance Level")
alpha = 0.05 
print(f"Significance Level: α = {alpha}")

print(f"\n Step 3: Perform the T-Test")
pop1 = df_final[df_final['Pop'] == 'Vic']['skullw_no_outlier']
pop2 = df_final[df_final['Pop'] != 'Vic']['skullw_no_outlier']
t_stat, p_val = ttest_ind(pop1, pop2, equal_var=False)
print(f"T-statistic: {t_stat}, P-value: {p_val}")

print(f"\n Step 4: Make a Decision")
if p_val < alpha:
    print(f"✅ Reject the null hypothesis (p = {p_val:.4f}).")
    print("There is a statistically significant difference in skull width between Vic and other populations.")
else:
    print(f"❌ Fail to reject the null hypothesis (p = {p_val:.4f}).")
    print("There is no statistically significant difference in skull width between Vic and other populations.")

```

# Simple regression 1 (Relationship Between Head Length and Chest Measurement)

```python
import seaborn as sns
sns.pairplot(df_final)
```

```python
from statsmodels.formula.api import ols
ols1=df_final[['hdlngth_no_outlier','chest_no_outlier']]
model1 = ols(formula='hdlngth_no_outlier ~ chest_no_outlier', data=ols1).fit()
print("Simple Regression 1: Head length vs Chest")
print(model1.summary())
```

# Explaination for simple regression 1
Head Length = 80.397 + 0.468 × Chest

The coefficient for chest_no_outlier is 0.4677, and the intercept is 80.3968, which means that head length equals the intercept (80.3968) plus the slope (0.4677) times chest measurement. In other words, for every one-unit increase in chest size, the head length is expected to increase by approximately 0.468 units.

The p-value associated with the chest coefficient is 0.000, indicating a statistically significant relationship between chest size and head length. The R-squared value of 0.162 suggests that about 16.2% of the variation in head length can be explained by chest size. The 95% confidence interval for the slope (0.259 to 0.677) indicates that we can be 95% confident that the true effect of chest on head length lies within this range.

Conclusion:
There is a significant positive linear relationship between chest size and head length, meaning that specimens with larger chest measurements tend to have longer head lengths. The model is statistically significant, though it explains a modest portion of the variability in head length.

# Simple regression 3 (Relationship Between Ear Conch and Foot Length)

```python
ols3=df_final[['earconch_no_outlier','footlgth_no_outlier']]

model3 = ols(formula='earconch_no_outlier ~ footlgth_no_outlier', data=ols3).fit()

print("Simple Regression 3: earconch vs footlgth")
print(model3.summary())
```

# Interpretation
The simple linear regression between ear length (earconch_no_outlier) and foot length (footlgth_no_outlier) shows a strong and statistically significant positive relationship. The regression equation is:

Ear Length = -1.694 + 0.728 × Foot Length

This means that for every one-unit increase in foot length, the ear length is expected to increase by approximately 0.73 units. The slope coefficient (0.7277) is highly significant (p < 0.001), indicating strong evidence that foot length is an important predictor of ear length. The intercept (-1.6944) has no meaningful interpretation in this context, as a foot length of zero is unrealistic, but it defines the position of the regression line.

The model’s R-squared value of 0.595 indicates that about 59.5% of the variation in ear length can be explained by foot length. The adjusted R-squared value of 0.591 confirms that the model provides a good fit, even after accounting for the single predictor. The F-statistic (149.9) with a p-value < 0.001 further supports the model’s overall significance. The 95% confidence interval for the slope (0.610 to 0.846) suggests that the true increase in ear length for each additional unit of foot length likely falls within this range.

In conclusion, the analysis demonstrates a clear and significant positive linear relationship between foot length and ear length. Individuals with longer foot measurements tend to have longer ear lengths, and foot length alone explains nearly 60% of the variation in ear size, indicating a strong predictive relationship.
